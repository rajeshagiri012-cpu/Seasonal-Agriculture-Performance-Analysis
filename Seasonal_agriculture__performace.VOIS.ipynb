```python
1)problem statement:
Agricultural activities are influenced by seasonal variations in environmental conditions,
farming practices, resource availability and market conditions. As a result, agricultural
performance may differ from one season to another.

  However, raw agricultural data does not clearly explain how agricultural performance changes
  across seasons or what patterns can be observed in different seasonal conditions.
  The problem is to analyze the given agricultural dataset and investigate seasonal differences
  in agricultural performance by identifying meaningful patterns, trends, relationships and
  variations within the available data.
```

```python
#Datset Uploading With importing the libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("/content/seasonal_agriculture_performance_dataset (2).csv")
print("dataset loaded suceesfully uploadded");
```

```python
#Analysis of basic information about the dataset

df.head() # top 5  rows analyzed
df.tail() # last 5 rows analyzed
```

```python
# DATAset shape and
# this function return NUMBER oF ROWS
# this function return Number of colmns
print("rows:",df.shape[0])
print("column:",df.shape[1])
```

```python
#dataset Datatypes checks
# df.info:it return the non-null count ,dtype ,columns of dataset it contain
# catgorized the catgorial types and numerical types
df.info()
categorial=df.select_dtypes(include=['object']).columns  # dividing the datatypes into categorial
Numerical=df.select_dtypes(include=['int64','float64']).columns # dividing the datatypes into numerical types
print(categorial)
print(Numerical)
```

```python
# identifying the missing values
df.isnull()# it return the toatl null values whee true says it is null values false indicates value (non-null)
df.isnull().sum() # it returns the each column the null values
missing_percent = (df.isnull().sum() / len(df)) * 100
missing_percent[missing_percent > 0] # missing values percentages
missing_percent.dtypes

# HANDLING THE MISSING VALUES
# we can handle the missing values by mean,median,mode
# mean and median for numerical median used if outilers detected mean without Outliers
#mode for catgorial
missing_cols = ["Rainfall_mm",
                "Soil_Moisture_pct",
                "Yield_Tonnes_Ha"]

for col in missing_cols:
    df[col] = df[col].fillna(df[col].median())
print(df.isnull().sum())

```

```python
# identifying the duplicates in the dataset
# in this datset there are no  duplicates if extist it can handled by drop the rows

df.duplicated()
df.duplicated().sum()
df.drop_duplicates(inplace=True)
df.duplicated().sum()
df.shape
```

```python
# describeing  the statical analysis of datase


df.describe() # describe the satistical analysis of the numerical columns
df.describe(include="object") # describe the satsiacal analysis of catgorial columns
df.describe(include="all") # describe the satsiacal analysis of all columns
```

```python
#)outliers  detected  and handling by iqr (inter quatile range) method

numeric_cols = df.select_dtypes(include="number").columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    print(col, ":", outliers)


    print("Negative values:")# checking numerical values that can be negative or unamous value
print((df[numeric_cols] < 0).sum())



#upto here we have completed the data cleaning loading the datasets checking null values,handling outliers etcc ;
```

```python
# next is to eda (explortary data analysis ) to find the disributions of dataset
# lets start the catergorial distribution
# uni varient analysis : means it has only one varialble at a time
# this count plot is used for catagerial data


#Seasonal performance of crops

df['Season'].value_counts()

plt.figure(figsize=(8,5))

sns.countplot(data=df, x="Season")

plt.title("Distribution of Farms Across Seasons")
plt.xlabel("Season")
plt.ylabel("Number of Farms")

plt.show()
#===================================================================================
# distribution of crops also be categorial we can say rice is more than other crops

df["Crop"].value_counts()
plt.figure(figsize=(10,5))

sns.countplot(data=df, x="Crop")

plt.title("Distribution of Crops")
plt.xlabel("Crop")
plt.ylabel("Number of Farms")

plt.xticks(rotation=45)

plt.show()

#===========================================================
#
df["Irrigation_Method"].value_counts()
plt.figure(figsize=(8,5))

sns.countplot(data=df, x="Irrigation_Method")

plt.title("Distribution of Irrigation Methods")
plt.xlabel("Irrigation Method")
plt.ylabel("Number of Farms")

plt.xticks(rotation=45)

plt.show()
#===================================================================
# most of the irragation can be done in flood  and another irragtion methods follows

```

```python
# numerical distributions : to represent the numerical values spread we use histplot  and kde used for smoothing the graph
# we can represent the yelid distribution and profit distribution through histogram



plt.figure(figsize=(8,5))

sns.histplot(data=df, x="Yield_Tonnes_Ha", kde=True)

plt.title("Distribution of Crop Yield")
plt.xlabel("Yield (Tonnes/Ha)")
plt.ylabel("Frequency")

plt.show()
#============================================================================================

plt.figure(figsize=(8,5))

sns.histplot(data=df, x="Profit_INR", kde=True)

plt.title("Distribution of Agricultural Profit")
plt.xlabel("Profit (INR)")
plt.ylabel("Frequency")

plt.show()
#===================================================================
plt.figure(figsize=(8,5))

sns.histplot(data=df, x="Production_Tonnes", kde=True)

plt.title("Distribution of Agricultural Production")
plt.xlabel("Production (Tonnes)")
plt.ylabel("Frequency")

plt.show()
#====================================================================================
```

```python
#average yelid by seasonal performance
# below are visulaztaions that can be realted with season with yelid, production,profit etc
# we clearly see that  average yelied per seson is in kharif where more crops as been grown
#and zaid season  is lowest yeild that can produce
season_yield = df.groupby("Season")["Yield_Tonnes_Ha"].mean()

print(season_yield)

plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x="Season",
    y="Yield_Tonnes_Ha",
    estimator="mean"
)

plt.title("Average Crop Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield (Tonnes/Ha)")

plt.show()
#==============================================
df.groupby("Season")["Total_Cost_INR"].mean()
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x="Season",
    y="Total_Cost_INR",
    estimator="mean"
)

plt.title("Average Agricultural Cost by Season")
plt.xlabel("Season")
plt.ylabel("Average Cost (INR)")

plt.show()
```

```python
# here we use the average profit by each season
# and we use function like df.group by () and barplot which to repressent the visualtion

season_profit = df.groupby("Season")["Profit_INR"].mean()

print(season_profit)
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x="Season",
    y="Profit_INR",
    estimator="mean"
)

plt.title("Average Agricultural Profit by Season")
plt.xlabel("Season")
plt.ylabel("Average Profit (INR)")

plt.show()
```

```python
#satatical analysis for sesonal agriculture
# the below table consists the avg (mean )of  yelid,prodution ,profit ,rainfall,water used according to season
# from this tablewe can eaasily answer every question which is least /lowest of every column

df.groupby("Season")[
    ["Yield_Tonnes_Ha",
     "Production_Tonnes",
     "Profit_INR",
     "Rainfall_mm",
     "Water_Used_m3"]
].mean()
```

```python
##key finds and insights
#KEY FINDS
# not visually we can here we can speak with numbers and we can answer by statically from above table
#==============================================================================================================
#INSIGHTS
# Kharif season recorded the highest average yield of 5.64 tonnes/ha, followed by Rabi at 5.08 tonnes/ha and Zaid at 4.67 tonnes/ha.
#Kharif season achieved the highest average profit of ₹178,914.65, while Zaid recorded an average loss of ₹24,804.82.
# these are insights =================================================================================
yield_by_season = df.groupby("Season")["Yield_Tonnes_Ha"].mean()

print(yield_by_season)
print("Best season for yield:", yield_by_season.idxmax())
print("Average yield:", round(yield_by_season.max(), 2))


#====================================================================
profit_by_season = df.groupby("Season")["Profit_INR"].mean()

print(profit_by_season)
print("Best season for profit:", profit_by_season.idxmax())
print("Average profit:", round(profit_by_season.max(), 2))
#====================================================================
production_by_season = df.groupby("Season")["Production_Tonnes"].mean()

print(production_by_season)
print("Best season for production:", production_by_season.idxmax())
```

## Recommendations

1. Kharif season showed the highest average yield, production, and profit.
   Therefore, farming practices used during this season can be studied and
   considered for suitable agricultural planning.

2. Zaid season recorded a negative average profit.
   Its cost, revenue, and resource usage should be investigated to identify
   possible reasons for the loss.

3. Water, fertilizer, and pesticide usage should be monitored across seasons
   to improve resource efficiency.

4. Environmental conditions such as rainfall, temperature, and soil moisture
   should be considered while planning seasonal farming activities.

   

## Limitations

1. The analysis is based only on the available dataset.
2. The dataset may not represent all agricultural regions or farms.
3. The analysis identifies relationships but does not prove cause and effect.
4. Market prices and environmental conditions may vary over time.


## Conclusion

The analysis shows that agricultural performance varies across seasons.
Kharif recorded the highest average yield, production, and profit among the
three seasons, while Zaid showed comparatively lower performance and a
negative average profit. The analysis of environmental conditions, resource
usage, and relationships between variables helps identify factors associated
with seasonal agricultural performance. These findings can support better
seasonal planning and resource management.
